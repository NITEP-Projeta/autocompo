```python
import pandas as pd
import numpy as np
```

```python
arquivo = "PRO-009-25_REV 1 - PROPOSTA COMERCIAL.xlsx"
aba = "CPU  1 "
```

```python
meta = pd.read_excel(arquivo, sheet_name=aba, nrows=2, header=None)
```

```python
data = pd.read_excel(arquivo, sheet_name=aba, skiprows=5)
```

```python
data.columns = data.columns.map(str)
```

```python
lista = ['Unnamed: 0', 'Cód.', 'Discriminação', 'Unid', 'Unnamed: 123']
```

```python
for i in data.columns:
    if i in lista:
        data[i] = data[i].astype(str)
    else:
        data[i] = pd.to_numeric(data[i], errors='coerce').round(2)
```

```python
data = data.dropna(subset=['Unid', 'P.Unit'])
```

```python
data = data.fillna(0)
```

```python
colunaBase = ['Cód.', 'Discriminação', 'Unid', 'P.Unit']
```

```python
colunaInicial = 5
```

```python
colunaTotal = len(data.columns)
```

```python
numeroGrupos = (colunaTotal - colunaInicial) // 3
```

```python
tabela_composições = []
```

```python
for i in range(numeroGrupos):
    idx = colunaInicial + i * 3
    identificador = str(meta.iloc[0, idx]).strip()
    nome = str(meta.iloc[1, idx]).strip()

    colunaQuantidade = data.columns[idx]
    colunaFutiliz = data.columns[idx + 1]
    colunaCusto = data.columns[idx + 2]

    bloco = data[colunaBase + [colunaQuantidade, colunaFutiliz, colunaCusto]].copy()
    bloco.columns = colunaBase + ['Quant', 'Futiliz', 'Custo']
    bloco['Composição'] = nome
    bloco['Identificador'] = identificador

    bloco = bloco.dropna(subset=['Quant', 'Custo'], how='all')
    bloco = bloco[bloco['Quant'] != 0]
    bloco = bloco.sort_values(by='Custo', ascending=False)

    tabela_composições.append(bloco)
```

```python
dataFinal = pd.concat( tabela_composições, ignore_index=True)
```

```python
# Ordena colunas finais
dataFinal = dataFinal[
    ['Cód.', 'Discriminação', 'Unid', 'P.Unit', 'Quant', 'Futiliz', 'Custo', 'Composição', 'Identificador']
]
```
